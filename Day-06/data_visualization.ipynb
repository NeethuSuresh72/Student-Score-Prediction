```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("student_scores.csv")

df
```

## Line Chart

```python
plt.figure(figsize=(8,5))

plt.plot(df["Hours_Studied"], df["Final_Score"], marker="o")

plt.title("Hours Studied vs Final Score")
plt.xlabel("Hours Studied")
plt.ylabel("Final Score")

plt.grid(True)

plt.show()
```

## Bar Chart

```python
plt.figure(figsize=(8,5))

plt.bar(df["Name"], df["Final_Score"])

plt.title("Student Final Scores")
plt.xlabel("Student")
plt.ylabel("Final Score")

plt.show()
```

## Histogram

```python
plt.figure(figsize=(8,5))

plt.hist(df["Final_Score"], bins=5)

plt.title("Distribution of Final Scores")
plt.xlabel("Final Score")
plt.ylabel("Frequency")

plt.show()
```

## Scatter Plot

```python
plt.figure(figsize=(8,5))

plt.scatter(df["Hours_Studied"], df["Final_Score"])

plt.title("Hours Studied vs Final Score")
plt.xlabel("Hours Studied")
plt.ylabel("Final Score")

plt.show()
```

## Conclusion

Data visualization helps identify trends, relationships, and patterns in datasets. These visualizations are useful before building machine learning models.
