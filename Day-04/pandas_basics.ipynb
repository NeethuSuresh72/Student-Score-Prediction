```python
import pandas as pd
```

## Creating a Pandas Series

```python
marks = pd.Series([85, 90, 78, 92, 88])

print(marks)
```

## Creating a DataFrame

```python
data = {
    "Name": ["Anu", "Rahul", "Sneha", "Arjun", "Meera"],
    "Age": [20, 21, 19, 22, 20],
    "Marks": [85, 90, 78, 92, 88]
}

df = pd.DataFrame(data)

df
```

## Display First and Last Records

```python
print("First Three Rows")
display(df.head(3))

print("Last Two Rows")
display(df.tail(2))
```

## Data Information

```python
df.info()
```

## Statistical Summary

```python
df.describe()
```

## Selecting Columns

```python
df["Name"]
```

```python
df[["Name", "Marks"]]
```

## Filtering Data

```python
df[df["Marks"] > 85]
```

## Adding a New Column

```python
df["Result"] = ["Pass", "Pass", "Pass", "Pass", "Pass"]

df
```

## Basic Statistics

```python
print("Average Marks:", df["Marks"].mean())
print("Highest Marks:", df["Marks"].max())
print("Lowest Marks:", df["Marks"].min())
```

## Conclusion

Pandas makes it easy to organize, analyze, and manipulate structured data. It is one of the most important libraries used in Data Science and Machine Learning.
