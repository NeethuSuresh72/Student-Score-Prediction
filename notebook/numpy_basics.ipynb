```python
import numpy as np
```

## Creating NumPy Arrays

```python
array1 = np.array([10, 20, 30, 40, 50])

print("Array:")
print(array1)
```

## Two-Dimensional Array

```python
array2 = np.array([[1, 2, 3],
                   [4, 5, 6]])

print(array2)
```

## Array Attributes

```python
print("Shape:", array2.shape)
print("Dimensions:", array2.ndim)
print("Data Type:", array2.dtype)
print("Size:", array2.size)
```

## Array Operations

```python
a = np.array([1, 2, 3])

b = np.array([4, 5, 6])

print("Addition:", a + b)
print("Subtraction:", b - a)
print("Multiplication:", a * b)
print("Division:", b / a)
```

## Mathematical Functions

```python
numbers = np.array([5, 10, 15, 20])

print("Sum:", np.sum(numbers))
print("Mean:", np.mean(numbers))
print("Maximum:", np.max(numbers))
print("Minimum:", np.min(numbers))
```

## Creating Special Arrays

```python
print("Zeros Array")
print(np.zeros((2,3)))

print()

print("Ones Array")
print(np.ones((2,3)))

print()

print("Identity Matrix")
print(np.eye(3))
```

## Conclusion

NumPy provides powerful tools for working with numerical data efficiently. It is widely used in Data Science, Machine Learning, and Artificial Intelligence.
