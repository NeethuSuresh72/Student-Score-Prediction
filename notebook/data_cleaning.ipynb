```python
import pandas as pd
```

```python
df = pd.read_csv("student_scores.csv")

df
```

## Dataset Information

```python
df.info()
```

```python
df.describe()
```

## Checking Missing Values

```python
df.isnull().sum()
```

## Filling Missing Values

```python
df["Attendance"] = df["Attendance"].fillna(df["Attendance"].mean())

df["Previous_Score"] = df["Previous_Score"].fillna(df["Previous_Score"].mean())

df
```

## Checking for Duplicates

```python
df.duplicated().sum()
```

## Saving the Cleaned Dataset

```python
df.to_csv("cleaned_student_scores.csv", index=False)

print("Dataset cleaned and saved successfully.")
```

## Conclusion

Data cleaning is an important preprocessing step before building a Machine Learning model. Handling missing values and checking for duplicates improves data quality and model performance.
